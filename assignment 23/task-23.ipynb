Q.2.

```python
# Split features & target
X = data.iloc[:, :-1]
y = data.iloc[:, -1]
```

```python
# Clean data
data = data.dropna()
```

```python
# Convert categorical → numeric
X = pd.get_dummies(X, drop_first=True)
```

```python
# Ensure target is numeric (0/1)
y = pd.factorize(y)[0]
```

```python
# Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python

# scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Model
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
# Prediction
y_pred = model.predict(X_test)
```

```python

from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, average='macro'))
print("Recall:", recall_score(y_test, y_pred, average='macro'))
print("F1 Score:", f1_score(y_test, y_pred, average='macro'))
```

Q.3

```python
from sklearn.neighbors import KNeighborsClassifier
k_values = [3, 5, 7]
results = {}
for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    results[k] = acc
    print(f"K={k}, Accuracy={acc}")

best_k = max(results, key=results.get)
print("Best K:", best_k)
```

Q.4

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report
nb = GaussianNB()
nb.fit(X_train, y_train)
y_pred = nb.predict(X_test)
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

Q.5

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
lr_pred = lr.predict(X_test)

# KNN (Best k = 7)
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(X_train, y_train)
knn_pred = knn.predict(X_test)

# Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
nb_pred = nb.predict(X_test)
```

```python
comparison = pd.DataFrame({
    "Algorithm": ["Logistic Regression", "KNN", "Naive Bayes"],

    "Accuracy": [
        accuracy_score(y_test, lr_pred),
        accuracy_score(y_test, knn_pred),
        accuracy_score(y_test, nb_pred)
    ],

    "Precision": [
        precision_score(y_test, lr_pred, average='weighted'),
        precision_score(y_test, knn_pred, average='weighted'),
        precision_score(y_test, nb_pred, average='weighted')
    ],

    "Recall": [
        recall_score(y_test, lr_pred, average='weighted'),
        recall_score(y_test, knn_pred, average='weighted'),
        recall_score(y_test, nb_pred, average='weighted')
    ],

    "F1 Score": [
        f1_score(y_test, lr_pred, average='weighted'),
        f1_score(y_test, knn_pred, average='weighted'),
        f1_score(y_test, nb_pred, average='weighted')
    ]
})
```

```python
print(comparison)

best = comparison.loc[comparison["Accuracy"].idxmax()]
print("\nBest Algorithm:", best["Algorithm"])
print("Reason: It has the highest Accuracy.")
```
